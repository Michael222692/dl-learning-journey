```python
rng = np.random.default_rng(seed=42)

z = rng.random((5, 5)) * 100

minimum = z.min()
maximum = z.max()

normalized = (
    (z - minimum)
    / (maximum - minimum)
)

print("原矩阵：")
print(z)

print("\n归一化后的矩阵：")
print(normalized)

print("\n归一化后的最小值：", normalized.min())
print("归一化后的最大值：", normalized.max())
```

```python
rng = np.random.default_rng(
    seed=42
)

z = rng.random(10)

maximum_index = z.argmax()
maximum_value = z[maximum_index]

print("原数组：")
print(z)

print("\n最大值：", maximum_value)
print("最大值索引：", maximum_index)

z[maximum_index] = 0

print("\n修改后的数组：")
print(z)
```

```python
coordinate_dtype = np.dtype([
    ("x", np.float32),
    ("y", np.float32)
])

grid = np.zeros(
    (5, 5),
    dtype=coordinate_dtype
)

values = np.linspace(
    0,
    1,
    5
)

x_grid, y_grid = np.meshgrid(
    values,
    values
)

grid["x"] = x_grid
grid["y"] = y_grid

print(grid)
```

```python
integer_types = [
    np.uint8,
    np.uint16,
    np.int8,
    np.int32,
    np.int64
]

for dtype in integer_types:
    info = np.iinfo(dtype)

    print(
        np.dtype(dtype).name,
        "最小值：",
        info.min,
        "最大值：",
        info.max
    )
```

```python
float_types = [
    np.float32,
    np.float64
]

for dtype in float_types:
    info = np.finfo(dtype)

    print("\n类型：", np.dtype(dtype).name)
    print("最小有限值：", info.min)
    print("最大有限值：", info.max)
    print("机器精度eps：", info.eps)
    print("最小正规正数tiny：", info.tiny)
```

```python
z = np.zeros(
    (40, 40)
)

with np.printoptions(
    threshold=np.inf
):
    print(z)
```

```python
z = np.arange(100)

target = 36.7

distances = np.abs(
    z - target
)

closest_index = distances.argmin()
closest_value = z[closest_index]

print("目标值：", target)
print("距离数组：")
print(distances)

print("\n最近值索引：", closest_index)
print("最近值：", closest_value)
```
